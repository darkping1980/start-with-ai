### 방법 1. 스키마(원하는 형식)로 원하는 형식을 강제하기 
- response_format  에 원하는 형식을 지정 
- stric

```python
schema = {
    "type": "json_schema",
    "json_schema": {
        "name": "news_info",
        "strict": True,
        "schema": {
            "type": "object",
            "properties": {
                "카테고리": {"type": "string"},
                "핵심어": {"type": "array", "items": {"type": "string"}},
                "한줄요약": {"type": "string"},
            },
            "required": ["카테고리", "핵심어", "한줄요약"],
            "additionalProperties": False,
        },
    },
}
```

```python
response = client.chat.completions.create(
    model="gpt-5.6-luna", 
    response_format=schema,
    messages=[
        {"role":"user","content":f"다음기사에서 {df['본문'][1]}"}
    ]
)    

result = response.choices[0].message.content
print(result)
```

### Pydantic로  제어
- 모델이 원하는 구조를 파이썬 클래스처럼 정의해서 사용

```python
from pydantic import BaseModel, Field

class NewsInfo(BaseModel):
    카테고리 : str = Field(description="경제, IT과학, 정치 등")
    핵심어 : list[str] 
    한줄요약 : str
    뉴스기사완성도 : int = Field(description="1~10")
```

```python
response = client.chat.completions.parse(
    model="gpt-5.6-luna", 
    response_format=NewsInfo,
    messages=[
        {"role":"user","content":f"다음기사 분석 {df['본문'][1]}"}
    ]
)    

result = response.choices[0].message.content
print(result)
```

### 중첩 스키마
- pydantic 의 모델안에 다른 모델

```python
import json
from typing import Optional
# 전체 내용을 스키마로 정의
# 등장하는 인물 뽑아보기
class Person(BaseModel):
    name: str = Field(description="인물 이름")
    role : str = Field(description="기사에서 역할이나 직함")



class ArticleEntity(BaseModel):
    title : str = Field(description="전체 기사내용을 한줄로")
    people : list[Person] = Field(description="기사에 등장하는 인물")
    date : Optional[str] = Field(description="기사 속 사건 날짜")


response = client.chat.completions.parse(
    model="gpt-5.6-luna", 
    response_format=ArticleEntity,
    messages=[
        {"role": "user", "content" : f"다음 기사에서 정보를 추출해줘. {df['본문'][1]}"}
    ]
)
# print(response.choices[0].message.content)
result = response.choices[0].message.content
result_dic = json.loads(result)
result_dic

```
