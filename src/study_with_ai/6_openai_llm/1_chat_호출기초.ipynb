### LLM 대화 기본 구조
- messages : 대화 목록
    - ai 대화
    - 내 대화
    - ai 대화
    - 내 대화
    - 맥락을 보존하려면 이 대화 내용 전체가 들어간다.
    - 나중에는 최적화
- role : 역활
    - system : AI 의 성격 지정, 가장 원초적인 지시, 규칙 담김, 맨 앞에 딱 한번
    - user(human) : 사람이 하는 말
    - assistant(ai) : AI 가 답변한 말

```python
# 가장 간단한 대화 - USER 만 쓰
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()       # api 
client = OpenAI()
```

```python
response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {
            "role" : "user",
            "content" :"파이썬이 너무 재밌다. 다른 재밌는 것도 있으면 추천해줘, 3줄 내외로"
        },
    ]
)

print(response)
```

```python
print(response.choices[0].message.content)
```

```python
print(response.choices[0].finish_reason)
```

```python
print(response.usage.total_tokens)
```

### 대화 이어하기
- 지난 대화 기억을 하지 못합니다.
- 지난 대화는 리스트에 담아놔야

```python
messages = [
    {"role": "user", "content": "너는 하츄핑이니?"},
    {"role": "assistant", "content": "그래 나는 하츄핑이야"},
    {"role": "user", "content": "노래 불러줘"}
]

response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)

print(response.choices[0].message.content)

```

```python
messages.append({"role":"user","content":"너가 아까 부른 노래 영어로 번역해줘"})

response = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=messages
)

print(response.choices[0].message.content)
```

### 스트리밍
- 답이 다 만들어진 후 받지말고, 조각조각으로 만들어질

```python
stream = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role": "user", "content": "AI 활용방법을 배우면 좋은점은?"}
    ],
    stream=True
)

pieces = []

for chunk in stream:

    piece = chunk.choices[0].delta.content
    print(piece, end="", flush=True)
    pieces.append(piece)

```
