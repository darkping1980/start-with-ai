```python
# 가장 간단한 대화 - USER 만 쓰
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()       # api 
client = OpenAI()
```

```python
# 요청하는 함수를 만들어서 재사용 - 시스템 프롬프트 를 적용한 단일 턴 대화
def ask(user, system=None, model="gpt-5.6-luna"):
    messages = []       # 전체대화내용
    if system:
        messages.append({"role":"system","content":system})                             # 시스템
    messages.append({"role":"user","content":user})                                     # 유저
    response = client.chat.completions.create(model=model, messages=messages)           # 응답

    return response.choices[0].message.content   
```

```python
question = "간단한 이력서를 작성해줘"
result = ask(question)
print(result)
```

```python
# 파일 저장해서 확인해보기
with open("result.md", "w", encoding="utf-8-sig") as file:
    file.write(result)
```

### 예시를 보여주기 - few-shot
- 예시를 넣으면 더 좋은 경우

```python
# 리뷰 감정 평가 시켜보기

review = "라로슈포제 화장품이 기대했던 것보다는 훨씬 좋았어요. 피부 당김도 적어졌고, 피부에 광이 나는 느낌이 들어요."

result = ask(f"다음 리뷰의 강점을 판단해줘: {review}")
print(result)
```

```python
few_ex = f"""
다음 리뷰의 감정을 긍정/부정/중립 중 하나로만 답하세요

# 예시
## 리뷰1 
- 정말 최고에요! -> 긍정

## 리뷰2 
- 최악이에요 -> 부정

## 리뷰3 
- 고만고만해요 -> 중립

# 리뷰
{review}
"""

result = ask(few_ex)
print(result)

```

```python
template = """
당신은 {tone} 초보자에게 기술을 쉽게 설명하고 분석하는 사람입니다.
사용자가 입력한 기술이나 개념을 초보자도 이해할 수 있도록 설명해 주세요.

설명 방식
- 어려운 전문 용어는 가능한 쉽게 풀어서 설명
- 전문 용어를 사용해야 한다면 바로 뒤에 쉬운 뜻을 함께 설명
- 먼저 이 기술이 무엇인지 간단하게 설명
- 이 기술이 왜 필요한지 설명
- 어떻게 동작하는지 쉽게 설명
- 어디에 사용하는 기술인지 설명
- 장점과 단점을 함께 설명
- 실제 사용 예시를 하나 이상 포함
- 초보자가 헷갈릴 만한 부분도 설명
- 이 기술과 함께 자주 사용되는 관련 기술도 소개
- 관련 기술과 현재 기술의 차이점이나 관계도 간단히 설명

설명 순서
1. 한 줄 요약
2. 기술이 무엇인지
3. 왜 필요한지
4. 어떻게 동작하는지
5. 어디에 사용하는지
6. 장점과 단점
7. 실제 사용 예시
8. 관련 기술
9. 관련 기술과의 차이점 또는 관계
10. 초보자가 기억하면 좋은 핵심 내용

출력 조건
- 마크다운 문법을 사용하지 않음
- #, ##, *, -, ** 같은 마크다운 기호를 사용하지 않음
- 너무 어려운 표현은 피함
- 설명은 자연스러운 문장으로 작성
- 초보자가 이해할 수 있는 수준으로 작성
"""

prompt = template.format(tone="친절한")
print(prompt)
```

```python
result = ask(user="자바에 대해서.",system=prompt)
print(result)
```
