### 벡터 DB
- 문서가 많을 때 대량 검색을 알아서 해주는 벡터 DB 구축
- 인덱싱(색인), 100만개 정도 되더라도 검색하고 계산이 거의 바로 된다. 
- 메타데이터(카테고리, 날짜 등)를 바탕으로 검색이 가능

Qeustion(질문) -> Retreiver(검색기)-벡터DB -> Generator(응답생성) 

- 근사 검색(ANN)
- HNSW
    - 벡터들을 여러 층의 그래프로 연결해두고, 위에서 대충 방향 잡아서 아래로 내려가면서 데이터를 훑는 방식

### 벡터 DB 종류
- ChromaDB
- FAISS(옛날 로컬 벡터 DB)
- Qdrant
- Pinecone
- pgvector - 기존의 postgreSQL 에 플러그인만 추가하면 사용가능
- neo4j(그래프 DB) - 그냥 벡터 DB 도 사용
- milvus - 데이터가 엄청 많을때

### ChromaDB 컬렉션 만들기
문서를 담는 단위 -> 이름 적어주고, sheet 개념이라고 생각
1. 임베딩 함수가 뭔지

```python
import os
from chromadb.utils.embedding_functions import OpenAIEmbeddingFunction  # chromaDB 에서 Openai 임베딩 모델
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()   # .env 파일에 저장되어있는 api 키 가져오기
client = OpenAI()
```

```python
embeddings = OpenAIEmbeddingFunction(
    model_name="text-embedding-3-small"
)

embed_client = chromadb.Client()

# collection 
collection = embed_client.get_or_create_collection(
    name="news_collection",
    embedding_function=embeddings,
    configuration={'hnsw' : {"space" : "cosine"}}
)
```

```python
print(collection.name)
```

```python
ids = df.index.astype(str).tolist()
metadata = df[['제목', '카테고리', '출처URL']].to_dict('records')
metadata
```

```python
collection.count()
```

```python
collection.upsert(ids=ids, documents=docs, metadatas=metadata)
collection.count()
```

```python
docs[:20]
```

```python
# 검색기
question = "대출 정보에 대해서 알려줘"

result = collection.query(query_texts=[question], n_results=5)
result
```

### Qdrant
- 방식은 거의 chromadb

```python
from qdrant_client import QdrantClient
from qdrant_client.models import Distance, VectorParams, PointStruct
```

```python
qdrant = QdrantClient(path="./qdrant_news_db")
qdrant
```

```python
qdrant.create_collection(
    collection_name="news_qdrant",
    vectors_config=VectorParams(size=1536, distance=Distance.COSINE)
)
```

```python
client = OpenAI()

response = client.embeddings.create(
    model="text-embedding-3-small",
    input=docs
)

response.data
```

```python

points = []
for i in range(len(docs)):
    points.append(PointStruct(
        id=i,
        vector=response.data[i].embedding,
        payload={'제목':df['제목'][i],'카테고리': df['카테고리'][i]}
    ))

points    
```

```python
qdrant.upsert(collection_name='news_qdrant', points=points)
```

```python
import numpy as np

# 임베딩 함수 만들기
def embed(texts):
    """텍스트 목록을 받아서 벡터 배열로 변환하는 함수. openai embedding 을 사용해서 X 1536 차원으로 변화"""

    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=texts
    )

    return np.array([item.embedding for item in response.data])
```

```python
question = "대출 정보에 대해서 알려줘"

query_vec = embed([question])[0]

result = qdrant.query_points(
    collection_name="news_qdrant",
    query = query_vec.tolist(),
    limit=5
)
print(result.points)
```

```python
for point in result.points:
    print(point.id)
    print(docs[point.id])
    print(point.payload)
```
