### 공공 데이터 API 사용해보기

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
service_key = os.getenv("SERVICE_KEY")
```

```python
import requests

url = "https://api.kcisa.kr/openapi/service/rest/convergence2019/getConver10"

headers = {
    "Accept" : "application/json"
}

params = {
    "serviceKey":service_key,
    "numOfRows":100,
    "pageNo":1,
    #"keyword":"축제",
    "where":"서울"  
}

response = requests.get(url, 
                        params=params,
                        headers=headers
                        )



print(response.status_code)

```

```python
print(response.text)
```

```python
data = response.json()  # 파이썬에서 사용가능한 형태로 변환
data['response']['body']['items']['item']
```

```python
len(data['response']['body']['items']['item'])
```

```python
# for 문으로 짜보기 

url = "https://api.kcisa.kr/openapi/service/rest/convergence2019/getConver10"

headers = {
    "Accept" : "application/json"
}

# 데이터를 담을 리스트를 만들어 봅시다.
total_data = []

# pageNo 를 1 ~ 5
for pageNo in range(1,6):
    params = {
        "serviceKey":service_key,
        "numOfRows":10,
        "pageNo":1,
        #"keyword":"축제",
        "where":"서울"  
    }

    response = requests.get(url, 
                            params=params,
                            headers=headers
                            )

    print(response.status_code)

    data = response.json()  # 파이썬에서 사용가능한 형태로 변환
    total_data += data['response']['body']['items']['item']
```

```python
len(total_data)
```

```python
# 12:50 분까지 자유롭게
# 1. 키워드도 바꿔보세요
# 2. 위치도 바꿔보세요
# 3. page, row 바꿔서 데이터 수집해보세요


url = "https://api.kcisa.kr/openapi/service/rest/convergence2019/getConver10"

headers = {
    "Accept" : "application/json"
}

# 데이터를 담을 리스트를 만들어 봅시다.
total_data = []

# pageNo 를 1 ~ 5
for pageNo in range(1,6):
    params = {
        "serviceKey":service_key,
        "numOfRows":5,
        "pageNo":1,
        #"keyword":"축제",
        "where":"서울"  
    }

    response = requests.get(url, 
                            params=params,
                            headers=headers
                            )

    print(response.status_code)

    data = response.json()  # 파이썬에서 사용가능한 형태로 변환
    total_data += data['response']['body']['items']['item']
```
