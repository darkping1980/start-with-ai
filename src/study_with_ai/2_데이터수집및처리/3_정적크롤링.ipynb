### 정적 크롤링
- API 가 없을 때 사람이 보는 웹페이지에서 필요한 부분을 프로그램으로 긁어 오는 방법
- 페이지 형태가 정적 페이지일 때(처음 로딩시 데이터가 다 로드 되는 페이지)

--- 
- HTML : 페이지의 구조를 정하는 역할
- CSS : 디자인을 정하는 역할 - 색깔, 크기, 레이아웃, 글꼴, 여백
- Javascript : 동작에 관여하는 역할 - 버튼 클릭, 입력, 검증

----
- HTML 파트만 보면 될 수 있다:내용은 여기에 있다
- 크롤링은 HTML 구조만 잘 읽으면 된다

태그로 구조를 만듭니다.
시작태그와 종료태그가 짝을 이루는데

```python
from bs4 import BeautifulSoup
import requests

name = "조승우"
url = f"https://ko.wikipedia.org/wiki/{name}"

# 위키백과가 기본 requests 요청을 봇으로 판단해 403(접근 거부)을 줄 수 있어요. 요청에 User-Agent(어떤 프로그램이 접속했는지 알려주는 정보)를 넣어 보세요.

headers = {
    "User-Agent": "StudyPythonBot/1.0 (contact: example@example.com)"
}


# 요청하기
response = requests.get(url, headers=headers, timeout=10)

# 상태코드
status = response.status_code
print(status)
```

```python
html = response.text
print(html)
```

```python
# html 의 내용을 가져오기 좋게 해주는 라이브러리가 -> BeautifulSoup 이다.
soap = BeautifulSoup(html, "html.parser")

# 제목 추출해보기
title = soap.select_one("#firstHeading")
title

history = soap.select_one("#mwFw")
history.text


history = soap.select_one("#mwFw")
print(history.text)

```

```python
section_data = soap.select("section")
section_data
```

```python
len(section_data)
```

```python
for i in section_data:
    print(i.text)
```

- .find : 조건에 맞는 첫번째 요소
- .find_all : 조건에 맞는 요소(리스트로 반환)
- .select_one : CSS 선택자로 첫번째 요소
- .select : CSS 선택자로 모든 요소(리스트)
- .get_text : 요소의 텍스트만 추출
- .get("속성명"):특정 속성값 추출
