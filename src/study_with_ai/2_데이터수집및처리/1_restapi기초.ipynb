### 1.API 와 HTTP 정리
- API 는 프로그램끼리 데이터를 주고 받는 약속
예) 식당에서 주방들어가서 주문 x -> 테이블의 태블릿세서 주문하면(주문방식이 공통:API) -> 음식이 옴(데이터받기)
---
- HTTP : 프로그램이 서버에 요청을 보내는 규칙
- GET : 데이터를 조회할때
- POST : **데이터** 를 전송할때

### 2. requests 외부 라이브러리로 GET 요청
- 설치 : uv add requests
- 주소(URL) 에 GET 요청을 보내면 JSON 으로 답이 온다

```python
import requests # 모듈불러오기 - 누군가가 만든 모듈 -> 사전 설치(외부 모듈)

url = "https://jsonplaceholder.typicode.com/todos/2"
res = requests.get(url)                                 # GET 방식으로 요청 - res 에 담음
status = res.status_code                                # 상태코드 - 200   
data = res.json()                                       # 실제 본문 데이터 -> json 형태로 변환

print("상태코드:",status)
print("실제데이터:",data)
```

```python
# 테스트 한번더 - params
url = "https://jsonplaceholder.typicode.com/comments"
res = requests.get(url,params={"postId":1})                                 # GET 방식으로 요청 - res 에 담음
status = res.status_code                                # 상태코드 - 200   
data = res.json()                                       # 실제 본문 데이터 -> json 형태로 변환

print("상태코드:",status)
print("실제데이터:",data)
```

```python
# 연습하기 - boody 내용만 담긴 리스트를 만들어 보세요
# 연습하기 - 2. https://jsonplaceholder.typicode.com/todos 주소에서 , userId가 1 인 사람의 타이틀만 저장해오기  
# 연습하기 - 3. 2의 데이터 중에서 타이틀들만 저장해보기

titles = []

url = "https://jsonplaceholder.typicode.com/todos"
res = requests.get(url,params={"userId":1})
data_list = res.json()


for data in data_list:
    titles.append(data["title"])      

print(*titles, sep="\n")
```
