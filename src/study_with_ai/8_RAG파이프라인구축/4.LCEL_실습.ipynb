### LCEL - 랭체인 문법
- Model, Prompt, Output, parser
- 각각을 부품처럼 갈아 끼우기 좋게 체인으로 연결해서 사용
- 실행 -> invoke

- Model : LLM 은 언제든지 바뀔수 있다. 
    - openai, gemini, claude, 
    - 로컬모델 : gemma, qwen
    - 한국모델 : LG 엑시원, KT 마음, SK A.X(에이닷), NC(바르코)
- Prompt : 프롬프트를 템플릿으로 관리(파일로 저장해서 불러와서 관리 가능) 
- parser : 답을 원하는 형식으로 다듬는 부분
- chain : chain = prompt | model | parser (| 로 연결을 하면 하나의 파이프라인이 구성)       

- Langchain : 각각 부품화 해서 조립하기 쉽게 해주는 라이브러리다.

### 모델 파트

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="openai:gpt-5.6-luna",
    use_responses_api="True",
    reasoning={'effort':'none'}
)

# 가장 단순한 model 호출
result = model.invoke("파이썬 에 대해서 알려줘")
print(result.text)
```

### 프롬프트 설정 - 부품이라고 생각

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages({
    ('system',"당신은 번역가야. {language} 로 아래 내용을 번역해주세요"),
    ('human','{question}')
})


tmp_prompt = prompt.invoke({"language":"영어","question":"나는 금요일 이라서 너무 좋다. 주말에 공부를 더 할수 있어서"})
```

```python
chain = prompt | model
result = chain.invoke({"language":"영어","question":"나는 금요일 이라서 너무 좋다. 주말에 공부를 더 할수 있어서"})

print(result.text)
```

### 출력형식

```python
from langchain_core.output_parsers import StrOutputParser, ListOutputParser, CommaSeparatedListOutputParser

chain = prompt | model | StrOutputParser()
result = chain.invoke({"language":"영어","question":"나는 금요일 이라서 너무 좋다. 주말에 공부를 더 할수 있어서"})

print(result)
```

```python
from langchain_core.output_parsers import StrOutputParser, ListOutputParser, CommaSeparatedListOutputParser

chain = prompt | model | CommaSeparatedListOutputParser()
result = chain.invoke({"language":"영어","question":"주말에 할 일등을 추천해줘, 리스트 형식으로 줘"})

print(result)
```

```python

prompt = ChatPromptTemplate.from_messages({
    ('system',"당신은 번역가야. {language} 로 아래 내용에 답해주세요"),
    ('human','{question}')
})


chain = prompt | model | CommaSeparatedListOutputParser()
result = chain.invoke({"language":"한국어","question":"주말에 할 일등을 추천해줘, 리스트 형식으로 줘"})

print(result)
```

### 연습하기
1. 영화 추천 LCEL 
2. 사용자 기분 에 맞는 LCEL
3. 캐릭터 설정해서 캐릭터의 톤, 말투에 맞춰서 답변하는 체인 만들기 

```python
# 3. 캐릭터 설정해서 캐릭터의 톤, 말투에 맞춰서 답변하는 체인 만들기 

prompt = ChatPromptTemplate.from_messages({
    ('system',"당신은 {character} 임  아래 내용에 답해주세요"),
    ('human','{question}')
})

# tmp_prompt =  prompt.invoke({"character":"스파이더맨","question":"강도가 나타났어 해치워줘"})
# print(tmp_prompt)


chain = prompt | model | StrOutputParser()
result = chain.invoke({"character":"매그니토","question":"강도가 나타났어 너의 초능력 으로 날 구해줘"})

print(result)
```

```python
# 영화 추천 LCEL 

prompt = ChatPromptTemplate.from_messages({
    ('system',"당신은 영화광임 아래 내용에 답해주세요"),
    ('human','{question}')
})

# tmp_prompt =  prompt.invoke({"character":"스파이더맨","question":"강도가 나타났어 해치워줘"})
# print(tmp_prompt)


chain = prompt | model | CommaSeparatedListOutputParser()
result = chain.invoke({"question":"재밌는 최신 영화 추천해 주세요"})


print(result)
```

```python
# 사용자 기분 에 맞는 LCEL

prompt = ChatPromptTemplate.from_messages({
    ('system',"현재 내 기분은 {status} 이런데 이거에 맞게 대답해줘"),
    ('human','{question}')
})

# tmp_prompt =  prompt.invoke({"character":"스파이더맨","question":"강도가 나타났어 해치워줘"})
# print(tmp_prompt)


chain = prompt | model | StrOutputParser()
result = chain.invoke({"status":"신남","question":"노래 추천 해줘"})

print(result)
```

### 구조화된 출력
- 예전에는 

```python
from pydantic import BaseModel, Field


class Place(BaseModel):
    name : str = Field(description="추천 여행 장소명") 
    reason : str = Field(description="추천 이유")
    date : str = Field(description="추천 여행일")


structured_model = model.with_structured_output(Place, method="json_schema")
response = structured_model.invoke("주말에 갈만한 여행지 추천해줘")
print(response)
```
