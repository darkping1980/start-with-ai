```python
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import List

class AnswerFormat(BaseModel):
    answer : str = Field(description="질문에 해당하는 답변")
    recommend : List[str] = Field(description="내일 할일 추천 목록")

parser = PydanticOutputParser(pydantic_object=AnswerFormat)

prompt_test = ChatPromptTemplate.from_messages([
    ('system','너는 {character}야. 말투도 {character} 말투로 답변해. {format} 에 맞춰서 답변하도록 해'),
    ('human',"친절하게 질문에 답을 하도록 해 {question}")
])

prompt = prompt_test.partial(
    format = parser.get_format_instructions()
)
prompt    
```

```python
chain = prompt | model | parser
result = chain.invoke({"character":"어린아이","question":"오늘 점심 추천"})
```

```python

# prompt = ChatPromptTemplate.from_template("너는 정책 안내 담당자야. 다음 질문에 {tone} 답변해라. {question}")

prompt_self_test = ChatPromptTemplate.from_template("너는 일반인 인데 {tone} 스타일 의 담당자 야, 뒤의 질문에 답변해줘 {question}")

prompt = prompt_self_test

# prompt : 질문형식, model: 언어모델,  StrOutputParser(): 출력형식
chain = prompt | model | StrOutputParser()

result = chain.invoke({"tone":"괴팍한","question":"오늘 점심 뭐 먹을까?"})
print(result)
```

```python

from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
from typing import List

class AnswerFormat(BaseModel):
    answer : str = Field(description="질문에 해당하는 답변")
    recommend : List[str] = Field(description="내일 할일 추천 목록")

parser = PydanticOutputParser(pydantic_object=AnswerFormat)

# 프롬프트 생성
# {tone}, {question}, {format} 값을 나중에 넣을 수 있도록 자리만 만들어 둠
prompt_self_test = ChatPromptTemplate.from_template(
    "너는 일반인 인데 {tone} 스타일 의 담당자 야, "
    "뒤의 질문에 답변해줘 {question} {format}"
)

# {format} 자리에 PydanticOutputParser가 요구하는 출력 형식을 미리 넣어둠
# 예: answer는 문자열, recommend는 문자열 리스트 형태로 출력하도록 모델에게 안내
prompt = prompt_self_test.partial(
    format=parser.get_format_instructions()
)

# prompt : 질문형식, model: 언어모델,  StrOutputParser(): 출력형식
chain = prompt | model | parser

result = chain.invoke({"tone":"괴팍한","question":"오늘 점심 뭐 먹을까?"})
print(result.answer)


```
