### 한국어 형태소 분석
- Kiwi
- 한국어는 영어와 달라서, 조사와 어미가 덕지 덕지 붙어 있다.
- 그래서 표현력이 남다르다.

```python
import pandas as pd

df = pd.read_csv("../data/11-1_뉴스정제.csv")
df[['제목','정제본문','카테고리']].head(3)
```

```python
from kiwipiepy import Kiwi

kiwi = Kiwi()
print(kiwi)
```

```python
# 경우 1 - 문장을 형태소로 나누기
sentence = "오늘은 날씨가 정말 좋아서 소풍을 꼭 가보고 싶다. 이 곳에 가면 좋을 것 같다."
result = kiwi.tokenize(sentence)
result
```

```python
for token in result:
    print(token.form, token.tag)
```

```python
nouns = []

for token in result:
    if token.tag.startswith("N"):
        nouns.append(token.form)  

nouns    
```

```python
verbs = []

for token in result:
    if token.tag.startswith("V"):
        verbs.append(token.form)  
verbs
```

```python
# 한 글자 명사는 걸러내기
nouns_filtered = []
for noun in nouns:
    if len(noun) > 1:
        nouns_filtered.append(noun)
nouns_filtered

```

```python
# 명사 추출 함수 만들기
def extract_nouns(text):
    nouns = []
    # 토크나이저 적용
    result = kiwi.tokenize(text)
    for token in result:
        if token.tag.startswith("N"):   # N 으로 시작하는 글자가 명사
            nouns.append(token.form)


    filtered_nouns = []
    for noun in nouns:
        if len(noun) > 1:
            filtered_nouns.append(noun)

    return 
    
```

```python
df['명사'] = df['정제본문'].apply(extract_nouns)
df[['정제본문','명사']].head(3)
```
