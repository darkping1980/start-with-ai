### 1. 임베딩
- 텍스트를 문장 단위로 벡터 : [0.25, 01.12, 0.11] 이런식으로 변환
- 검색, 추천시스템, RAG(챗봇), 분류 -> 이 벡터 변환 방식을 씁니다.

### 2. 임베팅 모델 준비
- 한국어에 강한 최신 모델 - 작고 빠른 걸로 - CPU 도 가능한

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("dragonkue/multilingual-e5-small-ko-v2")
print("모델 준비 완료")
```

```python
sentence = [
    "점심 메뉴로 돈까스를 추천합니다.",
    "오늘 점심은 김치찌개를 추천합니다.",
    "든든한 점심으로 제육볶음이 좋습니다.",
    "국물 있는 점심 메뉴로 순두부찌개를 추천합니다.",
    "간단한 점심으로 짜장면이 좋습니다.",
    "가볍게 먹고 싶다면 샌드위치를 추천합니다.",
    "한식 점심 메뉴로 비빔밥을 추천합니다.",
    "매콤한 점심 메뉴로 닭갈비가 좋습니다.",

    "저녁에 볼 만한 영화를 추천할 수 있습니다.",
    "파이썬은 예제를 직접 작성하면서 공부하는 것이 좋습니다.",
    "인공지능 개발은 다양한 모델을 실험해 볼 수 있는 분야입니다.",
    "주말 여행지로 가까운 바다나 산을 추천합니다."
]

passage_vecs = model.encode(
    ["passage: " + text for text in sentence],
    normalize_embeddings=True
)
```

```python
query = "오늘 점심 뭐 먹을까?"

query_vec = model.encode(
    "query: " + query,
    normalize_embeddings=True
)
```

```python
# 사용자가 입력한 질문과 의미가 비슷한 문장을 검색하는 함수
# query : 사용자가 입력한 검색 문장
# k     : 유사도가 높은 결과를 몇 개 출력할지 지정
def search_text_2(query, k=3):

    # 사용자의 질문에 "query: "를 붙여 임베딩 벡터로 변환
    # normalize_embeddings=True:
    # 벡터의 크기를 정규화해서 코사인 유사도 계산에 사용하기 좋게 만듦
    query_vec = model.encode(
        "query: " + query,
        normalize_embeddings=True
    )

    # 사용자의 질문 벡터와
    # 미리 만들어둔 답변 데이터(passage_vecs)의 유사도를 계산
    similarity = util.cos_sim(query_vec, passage_vecs)

    # 유사도가 높은 순서대로 정렬한 후
    # 상위 k개의 인덱스만 가져옴
    for i in similarity[0].argsort(descending=True)[:k]:

        # PyTorch Tensor 값을 일반 정수형으로 변환
        index = i.item()

        # 유사도 점수와 해당 문장을 출력
        print(
            similarity[0][index].item(),  # 유사도 점수
            sentence[index]               # 해당 인덱스의 원본 문장
        )
```

```python
search_text_2("매운거? ")
```
