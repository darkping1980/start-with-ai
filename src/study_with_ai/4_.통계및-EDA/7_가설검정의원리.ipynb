### t검정
- 가장 자주 만다는 질문 : 두 그룹의 평균이 다른가? 라는 질문
- 가장 기본적인 판단 기준

- 일표본 : 한 그룹의 평균이 특정 기준과 다른가?
    - 예) 평균 이용 시간이 70분과 다른가?
- 독립표본 : 서로 다른 두 그룹 비교
    - 남성과 여성
- 대응표본: 같은 대상의 두 조건 비교
    - 같은 상권의 주중 vs 주말    

### 1.일표본 t검정 -    

```python
t,p =  stats.ttest_1samp(df_gender['이용건수'], 40)
print("평균 이용 건수", df_gender['이용건수'].mean())
print(p)  
```

```python
df_gender.head(3)
```

```python
df_gender['이용시간(분)/건별'] = (df_gender['이용시간(분)'] / df_gender['이용건수']).round(2)
df_gender.head(3)
```

```python
male_time = df_gender.query(" 성별 == 'M' ")['이용시간(분)/건별']
print(male_time)

fmale_time = df_gender.query(" 성별 == 'F' ")['이용시간(분)/건별']
print(fmale_time)
```

```python
t, p = stats.ttest_ind(male_time, fmale_time, equal_var=False)

print(f"남자평균 {male_time.mean():.2f}") 
print(f"여자평균 {fmale_time.mean():.2f}") 
print("p-value",p)
```
