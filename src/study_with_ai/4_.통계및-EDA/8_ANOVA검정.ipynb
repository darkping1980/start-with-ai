### ANOVA 분석
- 세 그룹 이상의 평균을 한번에 비교하는 ANOVA
- 상권 데이터
- 상권 유형은 4가지 -> 데이터로 확인 해보고 시작

```python
import pandas as pd

df = pd.read_csv("../data/07-8_seoul_market_sales.csv")
df.head(3)
```

```python
df.describe()
```

```python
df.columns
```

```python
df['상권_구분_코드_명'].unique()
```

```python
groups = []

for name, g in df.groupby("상권_구분_코드_명"):
    #print(name, g)
    groups.append(g['당월_매출_금액'].values)
groups    
```

```python
len(groups)
```

```python
from scipy import stats
F,p = stats.f_oneway(*groups)
print("F 통계량",F)
print("p-value,",p)
```
