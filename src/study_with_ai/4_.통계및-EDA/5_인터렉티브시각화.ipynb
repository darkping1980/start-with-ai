```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv")
df.head()
```

```python
import plotly.express as px         # 실제로 가장 많이 사용 
import plotly.graph_objects as go   # 그래프 생성의 세세한 부분까지 다 제어가능
```

```python
import plotly.express as px

# 1. 차트 기본 생성 및 스타일 지정
fig = px.scatter(
    data_frame=df,
    x="공부시간",
    y="시험점수",
    trendline="ols",
    trendline_color_override="#E63946",  # 추세선 색상 (강렬한 레드)
    title="<b>공부시간과 시험점수의 상관관계</b>",  # 볼드 처리
    template="plotly_white",  # 깔끔한 흰색 배경 테마
)

# 2. 마커(점) 스타일 변경
fig.update_traces(
    marker=dict(
        size=10,  # 점 크기 확대
        color="#2A9D8F",  # 세련된 청록색
        opacity=0.8,  # 살짝 투명하게
        line=dict(width=1, color="#264653"),  # 점 테두리선
    ),
    selector=dict(mode="markers"),
)

# 3. 레이아웃 상세 커스텀
fig.update_layout(
    title=dict(
        font=dict(size=20, family="Malgun Gothic"),  # 제목 폰트 크기
        x=0.5,  # 제목 중앙 정렬
        xanchor="center",
    ),
    xaxis=dict(
        title="<b>공부 시간 (시간)</b>",
        gridcolor="#E5E5E5",  # 그리드선 연하게
        showline=True,
        linecolor="#333333",  # X축 기준선
    ),
    yaxis=dict(
        title="<b>시험 점수 (점)</b>",
        gridcolor="#E5E5E5",
        showline=True,
        linecolor="#333333",  # Y축 기준선
    ),
    width=700,  # 그래프 너비
    height=500,  # 그래프 높이
    font=dict(family="Malgun Gothic, AppleGothic, sans-serif"),  # 한글 폰트 적용
)

fig.show()
```

```python
mean = df.groupby("반", as_index=False)["시험점수"].mean()
print(mean)
```

```python
fig = px.bar(data_frame=mean, 
             x="반",
             y="시험점수",
             color="반",
             title="반별 평균 점수"
             )

fig.update_layout(width=800,height=500)
fig.show()
```

```python
# 파이차트
class_counts  = df['반'].value_counts().reset_index()
class_counts
```

```python
fig = px.pie(data_frame=class_counts,
             names='반',
             values="count")

fig.show()
```

```python
# 히스토그램과 박스플롯
fig = px.box(data_frame=df, 
             x="반",
             y="시험점수",
             color="반",
             points="all"
             )

fig.show()
```

```python
# 히스토그램 
fig = px.histogram(data_frame=df, 
                    x="반",

                    nbins=10
                    )
fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(rows=1, 
                    cols=2,
                    subplot_titles=["시험점수 분포","공부시간 분포"])

fig.add_trace(go.Histogram(x=df['시험점수']), row=1, col=1)
fig.add_trace(go.Histogram(x=df['공부시간']), row=1, col=2)

fig.show()
```
