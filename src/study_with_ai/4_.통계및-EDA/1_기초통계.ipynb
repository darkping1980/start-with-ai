```python
import pandas as pd

df = pd.read_csv("../data/07-1_study_habits.csv")
df.head()
```

### 2.중심- 평균,중앙값,최빈값

```python
print("평균:", df["시험점수"].mean() )        # 평균
print("중앙값:", df["시험점수"].median())     # 중앙값
print("최빈값:", df["시험점수"].mode())       # 최빈값
```

```python
mean_value = df["시험점수"].mean()
print("평균:", mean_value)
```

```python
median_value = df["시험점수"].median()
print(median_value)
```

```python
mode_value = df["시험점수"].mode()[0]
print(mode_value)
```

```python
# mean 직접 구하기
total_score = df["시험점수"].sum()
print(total_score)
```

```python
total_score / df['시험점수'].count()
```

```python
# 평균과 중앙이 다른경우 
df['용돈'].mean()
```

```python
df["용돈"].median()
```

### 4. 범위, 분산, 표준편차
- 범위 : 최댓값 ~ 최솟값
- 분산, 표준편차: 평균에서 얼마나 떨어져 있는지 보는 통곗값

```python
score = df["시험점수"]
print("최댓값", score.max())
print("최솟값", score.min())
print("범위", score.max() -score.min() )
```
