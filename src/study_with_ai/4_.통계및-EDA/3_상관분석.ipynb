### 상관분석
- 칼럼명 -> 변수 : 예) 수학점수, 용돈, 공부시간
- 산점도, 히트맵 등등

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv")
df.head()
```

### 2. 산점도로 관계보기
- 보려고하는 두 변수를 x,y 에 놓고 점을 찍어서 비교
- 공부시간,시험점수의 관

```python
sns.scatterplot(data=df, x="공부시간", y="시험점수")
plt.title("공부시간과 시험점수") # 인자를 하나만 전달
plt.show()
```

#### 3 상관계수


```python
corr = df["공부시간"].corr(df["시험점수"])
corr
```

```python
corr_matrix = df[["공부시간",'시험점수','용돈']].corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            vmin=-1)
plt.title("히트맵")
plt.show()
```
