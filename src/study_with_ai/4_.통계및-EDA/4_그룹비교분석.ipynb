```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("../data/07-1_study_habits.csv")
df.head()
```

```python
df.groupby("반")["시험점수"].mean().round(1)
```

```python
sns.barplot(data=df,x='반', y='시험점수')
plt.show()
```

```python
sns.boxplot(data=df, x='반',y='시험점수')
plt.show()
```

```python
sns.boxplot(data=df, x='반',y='시험점수', hue="성별")
plt.show()
```

```python
# 반별 공부시간 평균을 막대그래표로 그려보기 -3분- 


sns.barplot(data=df,x='반', y='공부시간',hue='성별', legend=False)
plt.show()



```
