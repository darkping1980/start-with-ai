```python
import pandas as pd

df = pd.read_csv("../data/07-1_study_habits.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
plt.rcParams["font.family"] = "Malgun Gothic"   # 한글 깨짐 방지(윈도우 기본 폰트)
plt.rcParams["axes.unicode_minus"] = False


df['시험점수'].hist(bins=15)
plt.title("시험 점수 분포")
plt.xlabel("점수")
plt.ylabel("학생수")
plt.show()
```

### 알 수 있는 부분
- 시험점수가 정규분포 형태다

```python
df['용돈'].hist(bins=10)
plt.title("용돈 분포")
plt.xlabel("용돈")
plt.ylabel("학생수")
plt.show()
```

```python
import seaborn as sns

sns.histplot(df['시험점수'], 
             bins=15, 
             kde=True)


```

```python
sns.boxplot(data=df)
plt.show()
```

```python
Q1 = df['용돈'].quantile(0.25)
Q3 = df['용돈'].quantile(0.75)
print(Q1, Q3)
IQR = Q3 - Q1
print("IQR",IQR)
```

```python
# 공부시간의 분포와 이상치를 보기 좋게 나타냅니다.
fig, ax = plt.subplots(figsize=(9, 4))

sns.boxplot(
    data=df,
    x='공부시간',
    ax=ax,
    color='#8EC5FC',       # 상자 색
    width=0.45,
    showmeans=True,       # 평균을 흰색 마름모로 표시
    meanprops={'marker': 'D', 'markerfacecolor': 'white',
               'markeredgecolor': '#1D4ED8', 'markersize': 7},
    medianprops={'color': '#1D4ED8', 'linewidth': 2},
    flierprops={'marker': 'o', 'markerfacecolor': '#EF476F',
                'markeredgecolor': 'white', 'markersize': 7}
)

ax.set_title('학생 공부시간 분포', fontsize=16, fontweight='bold', pad=16)
ax.set_xlabel('공부시간(시간)', fontsize=12)
ax.set_yticks([])
ax.grid(axis='x', color='#DDE3EA', linestyle='--', alpha=0.8)
sns.despine(left=True)
plt.tight_layout()
plt.show()
```
