### 파일 다루기
- 우리가 다루는 데이터는 코드상에 없어요
- 실제 데이터는 외부 별도 파일에 들어있다.
- 이 외부 별도 파일을 읽어오고, 수정하는 방법을 익혀야 한다.

- 지금까지 저장했던 리스트와 딕셔너리는 프로그램이 끝나면 사라짐
1. csv 파일 : 엑셀에서 다루기 쉬운 형태가 -> csv 파일
2. json 파일 : 웹에서 가져오는 데이터 형태
3. txt파일 : 로그 데이터 등등 일반적인 텍스트 형태

```python
# 파일 열어보기
data = ["이현근","김수연","송수림"]                       # 들어갈 데이터

file = open("text.txt","w", encoding="utf-8")          #  파일을 새로 만듭니다.
for i in data:
    file.write(i+"\n")                                 # 하나씩 기롤
file.close()                                           # 파일 닫기
```

```python
# 파일 만들기 - 2
data = ["이현근","김수연","송수림","김병주"]                       # 들어갈 데이터

with open("text.txt","w",encoding="utf-8") as file:
    for i in data:
        file.write(i + "\n")
```

```python
# 파일 읽어보기
# 1. 한번에 읽기    - read
# 2. 한줄씩 읽기    - readline : 반복문 
# 3. 한줄씩 읽어서 리스트로 읽기    - readlines 

# 1. 한번에 읽기
with open("text.txt","r",encoding="UTF-8") as file:
    text = file.read() # 전체 텍스트 파일 읽기
    print(text)
```

```python
# 2. 한줄씩 읽기    - readline : 반복문 
with open("text.txt","r",encoding="UTF-8") as file:
    while True:
        text = file.readline().strip()

        # 종료 조건을 지정해야줘야 무한반복 피함
        if(text == ""):     # 
            break           # 종료
        print(text)         # 읽은 데이터를 출력
```

```python
# 3. 한줄씩 읽어서 리스트로 읽기
with open("text.txt","r",encoding="UTF-8") as file:
    text = file.readlines()
    print(text)
```

```python
# 파일에 추가하기 a
with open("text.txt","a",encoding="UTF-8") as file:
    file.write("유종혁")
```

#### 예외처리
- 파일이 없는 경우
- 숫자가 아닌 값을 계산하라고 하는 경우
- 0으로 나누는 경우 등등
- 다양한 에러 상황을 처리할 수 있는 문법
    - try/except    사용법
    - else/finally  사용법
    - raise -> 일부러 에러를 띄우는 경우

```python
# 사용해보기
# try : 에러가 뜰수도 있는 경우 코드
# except : 에러 상황에 처리하는 코드

try:
    result = 10 /0
except Exception as error:
    print("에러입니다.",error)    
```

```python
try:
    result  = "abc" + 123
except Exception as error:
    print("에러입니다.",error)
```
