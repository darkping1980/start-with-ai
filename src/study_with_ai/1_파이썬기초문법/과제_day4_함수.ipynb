## 문제 6. 여러 값 반환 · ⭐⭐ 기본

숫자 리스트를 받아 **최솟값과 최댓값을 함께 return**하는 함수 `최소최대`를 만들고, 언패킹으로 받아 출력하세요.

```python
# TODO: def 최소최대(nums): return min(nums), max(nums)
# 그리고 최소, 최대 = 최소최대([...]) 로 받기
def 최소최대(nums):
    return min(nums), max(nums)

최소, 최대 = 최소최대([10,20,30,100])
print(f"최소:{최소},최대:{최대}")
```

## 문제 7. 개수가 정해지지 않은 인자 (*args) · ⭐⭐⭐ 응용

**여러 개의 숫자**를 받아 합계를 return하는 함수 `합계`를 `*나이`(또는 `*args`)로 만드세요.

```python
# TODO: def 합계(*nums): total 을 for 로 더해 return
def 합계(*nums):
    sum_num = 0
    for num in nums:
        sum_num += num

    return sum_num 

print(합계(10,100))        
```

## 문제 8. 키워드 인자 (**kwargs) · ⭐⭐⭐ 응용

이름=값 형태의 정보를 여러 개 받아 한 줄씩 출력하는 함수 `정보출력`을 `**kwargs`로 만드세요.

```python
# TODO: def 정보출력(**kwargs): for 키, 값 in kwargs.items(): 출력
def 정보출력(**kwargs):
    for 이름,값 in kwargs.items(): 
        print(f"{이름}={값}")

정보출력(이름="김병주",나이=46)                
```
