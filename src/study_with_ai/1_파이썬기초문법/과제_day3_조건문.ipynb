## 문제 11. 가위바위보 한 판 · ⭐⭐⭐ 응용

`나`와 `컴퓨터`의 손을 비교해 `이김`/`짐`/`비김`을 출력하세요.
(이기는 경우: 가위>보, 바위>가위, 보>바위)

```python
나 = "바위"
컴퓨터 = "가위"

이기는매핑 = {
                "가위":"보", 
                "바위":"가위",
                "보":"바위"
            }


if( 나 == 컴퓨터 ):
    print("비김")
elif( 이기는매핑[나] == 컴퓨터 ):
    print("이김")
else:
    print("짐")    



# TODO: 같으면 비김, 이기는 3가지면 이김, 나머지 짐
```

## 도전 1. 미니 계산기 · ⭐⭐⭐⭐ 도전

`연산`이 `"+"`,`"-"`,`"*"`,`"/"` 중 무엇이냐에 따라 `a`,`b`를 계산해 출력하세요.

```python
a = 12
b = 4
연산 = "*"

결과 = None

if(연산 == "+"):
    결과 = a + b
elif(연산 == "-"):
    결과 = a - b
elif(연산 == "*"):            
    결과 = a * b
elif(연산 == "/"):
    결과 = a / b    

# TODO: 연산 문자에 따라 if/elif 로 계산


print("결과:", 결과)
# 예상: 결과: 48
```

## 도전 2. 놀이기구 탑승 · ⭐⭐⭐⭐ 도전

`키`가 140 이상 **그리고** `나이`가 10 이상이면 `탑승 가능`.
아니면 이유를 안내: 키가 작으면 `키가 작아요`, 그 외엔 `나이가 어려요`.

```python
키 = 135
나이 = 12

if ( (키 > 140) and (나이 > 10) ):
    print("탑승 가능")
else:
    if(키 <= 140):
        print("키가 작아요")    
    elif(나이 <= 10):
        print("나이가 어려요")    


# TODO: and 로 둘 다 만족하면 탑승 가능, 아니면 이유 안내
```
