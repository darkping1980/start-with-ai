## 문제 15. 날짜 문자열 재조립 · ⭐⭐⭐ 응용

`"2026-08-31"`을 `split`으로 나눈 뒤 `"2026년 8월 31일"`로 만들어 출력하세요. (`int()`로 `08`→`8`)

```python
날짜 = "2026-08-31"

# TODO: split("-") 후 int 로 바꿔 f-string
parts = 날짜.split("-")
결과 = f"{parts[0]}년 {int(parts[1])}월 {parts[2]}일"
print(결과)
# 예상: 2026년 8월 31일
```

## 문제 16. 성적표 종합 · ⭐⭐⭐ 응용

점수에서 `영어`→95 수정, `과학`(80) 추가, **총점·평균·최고점**을 출력하세요. (`max(...values())`)

```python
점수 = {"국어": 90, "영어": 85, "수학": 100}

# TODO
점수["영어"] = 95

점수["과학"] = 80  # 과학 추가: 
총점 = sum(점수.values())
평균 = sum(점수.values()) / len(점수.values())
최고 = max(점수.values())
print("총점:", 총점, "/ 평균:", 평균, "/ 최고:", 최고)
# 예상: 총점 365 / 평균 91.25 / 최고 100
```
