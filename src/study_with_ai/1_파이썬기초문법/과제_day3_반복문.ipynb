## 문제 2. 1부터 10까지 합 · ⭐ 기초

`for`로 **1~10의 합**을 구해 출력하세요.

```python
합 = 0
# TODO: for 로 1~10 을 더하기

for i in range(1,11):
    합 += i

print("합:", 합)
# 예상: 합: 55
```

## 문제 6. 점수 합계와 평균 · ⭐⭐ 기본

점수 리스트를 `for`로 더해 **합계와 평균**을 출력하세요.

```python
점수 = [88, 92, 76, 100, 64]
합 = 0
# TODO: for 로 합을 구하고, 평균 = 합 / 개수

for 현재점수 in 점수:
    합 += 현재점수

print("합계:", 합, "/ 평균:", 합 / len(점수))
# 예상: 합계 420 / 평균 84.0
```

## 문제 7. 성인 세기 · ⭐⭐ 기본

나이 리스트에서 **19세 이상이 몇 명**인지 세어 출력하세요. (for + if)

```python
나이 = [15, 22, 34, 18, 40, 12]
성인 = 0
# TODO: for 로 돌며 19 이상이면 성인 += 1
for 나이변수 in 나이:
    if(나이변수) > 19:
        성인 += 1


print("성인:", 성인, "명")
# 예상: 성인 3 명
```

## 문제 8. 짝수만 모으기 · ⭐⭐ 기본

숫자 리스트에서 **짝수만** 골라 새 리스트에 모아 출력하세요. (for + if + append)

```python
숫자 = [7, 4, 9, 2, 8, 5, 10]
짝수 = []
# TODO: for 로 돌며 짝수면 append

for 숫자체크 in 숫자:
    if( (숫자체크 % 2) == 0 ):
        짝수.append(숫자체크)


print("짝수:", 짝수)
# 예상: 짝수 [4, 2, 8, 10]
```

## 문제 10. 합격자 명단 · ⭐⭐⭐ 응용

점수 딕셔너리를 순회하며 **80점 이상 합격자 이름**만 출력하세요. (for + items + if)

```python
점수 = {"민수": 90, "지현": 55, "수호": 100, "예린": 88}
# TODO: for 이름, 점 in 점수.items(): 80 이상이면 이름 출력

for 이름, 점수 in 점수.items():
    if( 점수 > 80 ):
        print(이름)

```

## 도전 1. 369 게임 · ⭐⭐⭐⭐ 도전

1부터 20까지 세되, **3의 배수(3,6,9...)에서는 `"짝"`**, 아니면 숫자를 출력하세요.

```python
# TODO: for n in range(1, 21): 3의 배수면 "짝", 아니면 n
for i in range(1,21):
    if( i % 3 == 0):
        print("짝")
    else:
        print(i)            
```

## 도전 2. 구구단 전체 · ⭐⭐⭐⭐ 도전

2단부터 9단까지 모두 출력하세요. (반복문 안에 반복문 - 2중 for)

```python
# TODO: for 단 in range(2, 10): 안에서 for i in range(1, 10):
for 단 in range(2,10):
    for i in range(1, 10):
        print(f"{단} × {i} = { 단 * i } ")        

```
