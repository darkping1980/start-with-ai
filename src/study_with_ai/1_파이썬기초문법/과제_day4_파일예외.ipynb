## 문제 4. 여러 줄 저장 · ⭐⭐ 기본

이름 리스트 `["이현근", "김수연", "송수림"]`을 한 줄에 하나씩 `명단.txt`에 저장하세요. (반복문 + write)

```python
names = ["이현근", "김수연", "송수림"]
# TODO: with open("명단.txt","w",...) as f: for n in names: f.write(n + "\n")
with open("명단.txt","w", encoding="utf-8") as file:
    for name in names:
        file.write(name + "\n")

print("저장 완료")
```

## 문제 5. 0으로 나누기 막기 · ⭐⭐ 기본

```python
# TODO: try: 10/0  except ZeroDivisionError: 안내 출력
try:
    result = 10/0
except ZeroDivisionError:
    print("0으로 나눌 수 없어요")    
```
