```python
# 연습 - 딕셔너리로 총 지출내역을 저장해놓고, 입력시 전체 지출 금액을 출력하는 함수를 만들어봅시다.

out_list  = {"점심" : 300_000,
             "저녁" : 400_000, 
             "통신비" : 120_000, 
             "교통비": 80_000 
             }



# 함수 선언 시작
def expenses_func(out_list):
    expenses = 0                        # 총 지출 금액

    for i in out_list.values():         # 지출내역에서 금액부분만 가져와서     
        expenses += i
    print(expenses)    
# 함수 선언 끝

expenses_func(out_list)
```

```python
# bmi 구하는 함수 
def get_bmi(weight, height):
    bmi = weight / (height * height)
    print(bmi)

get_bmi(80,1.75)    # bmi 구하는 함수 호출
```

```python
count = 20
def calculate(a, b):
    global count
    count += 50
    return (count + a + b)
```

```python
calculate(30,40)
```

```python
def check_number(number):
    if number % 2 == 0:     # 2 로 나눠서 나머지가 0 -> 짝수다
        return "짝수"
    else:
        return "홀수"            
```

```python
check_number(10)
```

```python
# 주민등록
# 예) 901201-3413115 -> "남자"

def id_check(id_number):
    # 여기에 내용 넣어 보세요
    # print(id_number[7:8])
    if( int(id_number[7:8]) % 2 == 0):
        gender = "여자"
    else:
        gender = "남자"        
    
    return gender
```

```python
ret_gender = id_check("800114-1852211")
print(ret_gender)
```
