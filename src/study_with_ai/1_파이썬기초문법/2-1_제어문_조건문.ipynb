조건문

- if
- if ~ else
- if ~ elif ~ else

#### 조건문이 중요한 이유
- 지금까지 코드는 그냥 위에서 아래로 실행
- 특정 상황에 따라서 실행여부를 결정할 수 있도록 해야 할 때 - 조건문이 필요하다
- 예) 
    - 나이가 19살 이상이다 -> 성인, 아니다 -> 미성년자 출력
    - 점수가 90점 이상이다 -> A, 80점 이상이다 -> B
    - 물건 재고가 있을 경우 -> 주문 가능, 재고가 없을경우 -> 품절로 안내

- 다양한 갈림길에서 실행 여부를 정해주는 것이 if 조건 -> 조건문    

1. if - 조건

```python
start_time = 10

if start_time > 9:
    print("지각압니다.")
```

```python
item = 50000
bank_account = 300000
if item < bank_account:
    print("구매 가능")
```

#### 2. if ~ else

    조건이 맞을때 실행
    아닐때 실행

```python
age = 15

if age > 19:
    print("성인입니다.")
else: 
    print("미성년자입니다.")    
```

```python
number = 8

if number % 2 == 0: 
    print("짝수 입니다.")
else :
    print("홀수 입니다.")    
```

```python
weather = "비"

if weather  == "비":
    print("우산을 챙기세요")
else:
    print("우산 없어도 됩니다.")    
```

#### 3. if~ elif~ else
- 갈림길이 여러 개일 때
- 조건이 여러 개일 때

```python
score = 75

if score > 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print("학점은" + grade)   
```

중첩 if
    - 조건문안의 조건문

```python
## 만약에 나이가 20이상이고, 회원이 가능할때 입장 가능
age = 19
is_member = True

if age >= 20:
    if is_member == True:
        print("입장가능") 
    else:
        print("회원만 입장 가능")

else: 
    print("성인만 입장 가능")
```

```python
# 영화관 심야 관람
# 나이가 15세 이상일때 관람 가능 
# 나이가 15세 미만일때 보호자와 관람 가능
# 아닐때는 관람이 불가능하다

age = 14
is_bohoza = True

if age > 15:
    print("관람이 가능합니다.")
    if is_bohoza == True:
        print("보호자와 관람 가능합니다.")
    else :
        print("관람이 불가능합니다.")    
```

```python
# 게임 대회 참가
level, is_reg = input().split()
level = int(level)
is_reg = (is_reg == "True")

if level >= 10 :
    if is_reg == True:
        print("대회 참가 가능") 
    else:
        print("신청이 필요합니다.")    
else:    
    print("레벨이 부족합니다.")
```

```python
# 도서관 책 대출

overdue_books, is_library_member = input().split()

overdue_books = int(overdue_books)
is_library_member = (is_library_member == "True")

if is_library_member == True:
    if ( overdue_books > 0 ):
        print("연체 도서를 먼저 반납하세요")
    else:
        print("대출가능")
else:
    print("회원가입 이 필요합니다.")            
```
