각 문제의 **빈칸을 채워** 실행하고, 예상 결과가 나오면 성공입니다. 막히면 아래 「정답 예시」를 참고하세요.

## 문제 1

`for`와 `range`로 **1부터 10까지의 합**을 구해 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 합=0 으로 시작 -> for i in range(1,11): 합 += i
totalSum = 0

for i in range(1,11):
    totalSum += i

print(totalSum)
```

```python
# 정답 예시
합 = 0
for i in range(1, 11):
    합 += i
print("합:", 합)
# 예상: 합 55
```

## 문제 2

`while`로 5부터 1까지 **거꾸로** 출력하세요.

```python
# 여기에 작성하세요
# 힌트: n=5 -> while n >= 1: print(n); n -= 1
n = 5

while n >= 1:
    print(n)
    n -= 1
    

```

```python
# 정답 예시
n = 5
while n >= 1:
    print(n)
    n -= 1
# 예상: 5 4 3 2 1
```

## 문제 3

나이 리스트 `[15,22,34,18,40]`에서 **19세 이상이 몇 명**인지 세어 출력하세요.

```python
ageList = [15,22,34,18,40]
ageCheckCount = 0

for age in ageList :
    if( age > 19 ):
        ageCheckCount += 1

print(ageCheckCount)

# 여기에 작성하세요
# 힌트: 성인=0 -> for a in 나이: if a >= 19: 성인 += 1
```

```python
# 정답 예시
나이 = [15, 22, 34, 18, 40]
성인 = 0
for a in 나이:
    if a >= 19:
        성인 += 1
print("성인:", 성인, "명")
# 예상: 성인 3 명
```

## 문제 4

숫자 리스트 `[7,4,9,2,8,5,10]`에서 **짝수만** 골라 새 리스트에 모아 출력하세요.

```python
num_list = [7,4,9,2,8,5,10]
new_num_list = []

for num in num_list:
    if( num % 2 == 0):        
        new_num_list.append(num)

print(new_num_list)        

# 여기에 작성하세요
# 힌트: 빈 리스트 -> for -> if n % 2 == 0: append
```

```python
# 정답 예시
숫자 = [7, 4, 9, 2, 8, 5, 10]
짝수 = []
for n in 숫자:
    if n % 2 == 0:
        짝수.append(n)
print("짝수:", 짝수)
# 예상: 짝수 [4, 2, 8, 10]
```

## 문제 5

점수 딕셔너리 `{"민수":90,"지현":55,"수호":100,"예린":88}`에서 **80점 이상 합격자 이름**만 출력하세요.

```python
people_score_list = {"민수":90,"지현":55, "수호":100, "예린":88}

name = ''
score = 0


for name, score in people_score_list.items():
    if(score > 80):
        print("합격자 :", name)
    



# 여기에 작성하세요
# 힌트: for 이름, 점 in 점수.items(): if 점 >= 80: print(이름)
```

```python
# 정답 예시
점수 = {"민수": 90, "지현": 55, "수호": 100, "예린": 88}
for 이름, 점 in 점수.items():
    if 점 >= 80:
        print("합격:", 이름)
# 예상: 합격 민수 / 수호 / 예린
```

## 도전 · 구구단 & FizzBuzz (재미)

**(1) 구구단**: 3단을 `for`로 출력하세요. (`3 x 1 = 3` ... `3 x 9 = 27`)
**(2) FizzBuzz**: 1~15에서 3의 배수는 `Fizz`, 5의 배수는 `Buzz`, 둘 다면 `FizzBuzz`, 아니면 숫자를 출력하세요.

```python
for i in range(1,11):
    print(f"3 * {i} = {(3 * i)}")

# 여기에 작성하세요 (구구단 3단)
# 힌트: for i in range(1, 10): print(f"3 x {i} = {3*i}")
```

```python
# 정답 예시 (구구단 3단)
for i in range(1, 10):
    print(f"3 x {i} = {3 * i}")
# 예상: 3 x 1 = 3 ... 3 x 9 = 27
```

```python


# 여기에 작성하세요 (FizzBuzz 1~15)
# 힌트: for n in range(1,16): 3과 5 둘 다 -> FizzBuzz 먼저 검사
```

```python
# 정답 예시 (FizzBuzz)
for n in range(1, 16):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
# 예상: 1 2 Fizz 4 Buzz Fizz 7 8 Fizz Buzz 11 Fizz 13 14 FizzBuzz
```
