각 문제의 **빈칸을 채워** 실행하고, 예상 결과가 나오면 성공입니다. 막히면 아래 「정답 예시」를 참고하세요.

## 문제 1

이름 `유나`, 나이 `20`을 변수에 담고, f-문자열로 `유나님은 20세입니다`를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: name, age 변수에 담고 f"{name}님은 {age}세입니다"
```

```python
# 정답 예시
name = "유나"
age = 20
print(f"{name}님은 {age}세입니다")
# 예상: 유나님은 20세입니다
```

## 문제 2

17을 5로 나눈 **몫**과 **나머지**, 그리고 2의 **10제곱**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 몫 //, 나머지 %, 거듭제곱 **
```

```python
# 정답 예시
print("몫:", 17 // 5)
print("나머지:", 17 % 5)
print("2의 10제곱:", 2 ** 10)
# 예상: 몫 3 / 나머지 2 / 2의 10제곱 1024
```

## 문제 3

문자열 `"Python"`에서 **첫 글자**와 **마지막 글자**, 그리고 **앞 3글자**를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 인덱싱 [0], [-1] / 슬라이싱 [:3]
```

```python
# 정답 예시
s = "Python"
print("첫 글자:", s[0])
print("마지막:", s[-1])
print("앞 3글자:", s[:3])
# 예상: 첫 글자 P / 마지막 n / 앞 3글자 Pyt
```

## 문제 4

문자열 `"hello world"`를 **모두 대문자로** 바꾸고, 공백을 기준으로 **단어 리스트로** 나눠 출력하세요.

```python
# 여기에 작성하세요
# 힌트: upper() / split()
```

```python
# 정답 예시
s = "hello world"
print(s.upper())
print(s.split())
# 예상: HELLO WORLD / ['hello', 'world']
```

## 문제 5

문자열 `"25"`를 **정수로 바꿔** 5를 더한 값을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: int() 로 형변환 후 덧셈
```

```python
# 정답 예시
x = "25"
print(int(x) + 5)
# 예상: 30
```
