강의(02-5)의 if·elif·else를 직접 풀어봅니다. (반복문은 아직 안 배웠으니 쓰지 않습니다.)

각 문제의 **빈칸을 채워** 실행하고, 예상 결과가 나오면 성공입니다. 막히면 아래 「정답 예시」를 참고하세요.

## 문제 1

`나이 = 20` 일 때, 19 이상이면 `성인`, 아니면 `미성년자`를 출력하세요.

```python
# 여기에 작성하세요
# 힌트: if 나이 >= 19: ... else: ...
age = 20

if (age > 19):
    print("성인")
else:
    print("미성년자")     
```

```python
# 정답 예시
나이 = 20
if 나이 >= 19:
    print("성인")
else:
    print("미성년자")
# 예상: 성인
```

## 문제 2

`점수 = 82` 를 학점으로: 90이상 A, 80이상 B, 70이상 C, 나머지 F 를 출력하세요.

```python
score = 82

if( score >= 90 ):
    print("A")
elif( score >= 80 ):
    print("B")
elif( score >= 70 ):
    print("C")
else:
    print("F")            

# 여기에 작성하세요
# 힌트: if/elif/elif/else
```

```python
# 정답 예시
점수 = 82
if 점수 >= 90:
    print("A")
elif 점수 >= 80:
    print("B")
elif 점수 >= 70:
    print("C")
else:
    print("F")
# 예상: B
```

## 문제 3

`기온 = 28` 이 20 이상 30 미만이면 `쾌적`, 아니면 `주의`를 출력하세요.

```python
temperature = 20

if( (20 <= temperature) and ( temperature < 30 )  ):
    print("쾌적")
else:
    print("주의")


# 여기에 작성하세요
# 힌트: 논리 and 로 범위 조건
```

```python
# 정답 예시
기온 = 28
if 기온 >= 20 and 기온 < 30:
    print("쾌적")
else:
    print("주의")
# 예상: 쾌적
```

## 문제 4

점수 딕셔너리 `{"민수":90,"지현":55}`에서 `"민수"` 점수가 60 이상이면 `합격`, 아니면 `불합격`을 출력하세요.

```python
score = {"민수":90,"지현":55}

if (score["민수"] >= 60):
    print("합격")
else:
    print("불합격")    



# 여기에 작성하세요
# 힌트: 딕셔너리[키]로 값을 꺼내 if 로 판정
```

```python
# 정답 예시
점수 = {"민수": 90, "지현": 55}
if 점수["민수"] >= 60:
    print("민수: 합격")
else:
    print("민수: 불합격")
# 예상: 민수: 합격
```

## 문제 5

`돈 = 3000`, `가격 = 4500` 일 때, 돈이 가격 이상이면 `구매 가능`, 아니면 `부족: 남은 금액`을 출력하세요.

```python
money = 3000
price = 4500

if( money > price ):
    print("구매가능")
else:
    print("부족: 남은 금액 ", price  - money)    

# 여기에 작성하세요
# 힌트: if 돈 >= 가격 / else 에서 가격-돈
```

```python
# 정답 예시
돈 = 3000
가격 = 4500
if 돈 >= 가격:
    print("구매 가능")
else:
    print("부족:", 가격 - 돈, "원")
# 예상: 부족: 1500 원
```
