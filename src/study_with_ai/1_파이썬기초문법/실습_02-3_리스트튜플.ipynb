## 문제 1

동물 리스트 `["사자","코끼리","펭귄","기린"]`에서 **두 번째**와 **마지막** 동물을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 인덱스 0부터. 마지막은 [-1]
animalList = ["사자","코끼리","펭귄","기린"]
print(animalList[1]) 
print(animalList[-1])
```

```python
# 정답 예시
동물 = ["사자", "코끼리", "펭귄", "기린"]
print("두 번째:", 동물[1])
print("마지막:", 동물[-1])
# 예상: 두 번째 코끼리 / 마지막 기린
```

## 문제 2

수 리스트 `[10,20,30,40,50]`에서 **가운데 세 개**(20,30,40)를 슬라이싱으로 출력하세요.

```python
# 여기에 작성하세요
# 힌트: [1:4]
numbers = [10,20,30,40,50]
print(numbers[1:4])
```

```python
# 정답 예시
수 = [10, 20, 30, 40, 50]
print(수[1:4])
# 예상: [20, 30, 40]
```

## 문제 3

가격 리스트 `[1200,3400,900]`에 **2500을 추가**하고, **개수·합계·평균**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: append 후 len·sum
priceList = [1200,3400,900]
priceList.append(2500)
print("개수",len(priceList))
print("합계",sum(priceList))
print("평균", sum(priceList) / len(priceList) )
```

```python
# 정답 예시
가격 = [1200, 3400, 900]
가격.append(2500)
print("개수:", len(가격))
print("합계:", sum(가격))
print("평균:", sum(가격) / len(가격))
# 예상: 개수 4 / 합계 8000 / 평균 2000.0
```

## 문제 4

점수 리스트 `[88,92,76,100,64]`의 **최고점·최저점**을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: max() / min()
score_list = [88,92,76,100,64]
print("최고점",max(score_list))
print("최저점",min(score_list))
```

```python
# 정답 예시
점수 = [88, 92, 76, 100, 64]
print("최고:", max(점수))
print("최저:", min(점수))
# 예상: 최고 100 / 최저 64
```

## 문제 5

튜플 `(2026, 3, 15)`을 `연`,`월`,`일` 세 변수에 나눠 담아 `2026년 3월 15일`을 출력하세요.

```python
# 여기에 작성하세요
# 힌트: 언패킹: 연, 월, 일 = 날짜
date = (2026, 3, 15)
year, month, day = date
print(f"{year}년 {month}월 {day}일")
```

```python
# 정답 예시
날짜 = (2026, 3, 15)
연, 월, 일 = 날짜
print(f"{연}년 {월}월 {일}일")
# 예상: 2026년 3월 15일
```
