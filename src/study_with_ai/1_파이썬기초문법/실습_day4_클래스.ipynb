## 문제 2. __init__ 로 초기화 · ⭐⭐ 기본

만들 때 `이름`·`나이`를 받는 클래스 `학생`을 `__init__`으로 만들고, 객체의 이름·나이를 출력하세요.

```python
# TODO: class 학생: def __init__(self, 이름, 나이): self.이름=... self.나이=...
class 학생:

    def __init__(self,이름,나이):
        self.이름 = 이름
        self.나이 = 나이

    def introduce(self):
        print(f" 이름은 {self.이름} 이고 나이는 {self.나이} 살 입니다. ")

student = 학생("김병주",46)
student.introduce()

```

```python
# 정답 예시
class 학생:
    def __init__(self, 이름, 나이):
        self.이름 = 이름
        self.나이 = 나이

s = 학생("김수연", 25)
print(s.이름, s.나이)
# 예상: 김수연 25
```

## 문제 3. 메서드 추가 · ⭐⭐ 기본

`학생` 클래스에 `소개` 메서드를 넣어 `"OO입니다. 25살이에요."`를 출력하게 하고, 호출하세요.

```python
# TODO: class 학생 안에 def 소개(self): print(...)

class 학생:

    def __init__(self,이름,나이):
        self.이름 = 이름
        self.나이 = 나이

    def introduce(self):
        print(f"  {self.이름} 입니다. {self.나이}살이에요.  ")

student = 학생("김병주",46)
student.introduce()

```

```python
# 정답 예시
class 학생:
    def __init__(self, 이름, 나이):
        self.이름 = 이름
        self.나이 = 나이
    def 소개(self):
        print(f"{self.이름}입니다. {self.나이}살이에요.")

s = 학생("김수연", 25)
s.소개()
# 예상: 김수연입니다. 25살이에요.
```

## 도전. 책 클래스 · ⭐⭐⭐ 도전

`제목`·`저자`를 받는 클래스 `책`을 만들고, `정보` 메서드로 `"제목 - 저자"`를 출력하게 하세요.

```python
# TODO: class 책: __init__(self, 제목, 저자) / 정보 메서드
class 책:
    def __init__(self,제목,저자):
        self.제목 = 제목
        self.저자 = 저자

    def 정보(self):
        print(f"{self.제목}-{self.저자}")


book = 책("허클베리핀","모름")
book.정보()

```

```python


# 정답 예시
class 책:
    def __init__(self, 제목, 저자):
        self.제목 = 제목
        self.저자 = 저자
    def 정보(self):
        print(f"{self.제목} - {self.저자}")

b = 책("파이썬 입문", "홍길동")
b.정보()
# 예상: 파이썬 입문 - 홍길동
```
