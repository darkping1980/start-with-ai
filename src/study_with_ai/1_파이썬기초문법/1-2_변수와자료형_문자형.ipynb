```python
product_name = "파이썬노트북"
first_letter = product_name[0:1]
print(first_letter)

last_letter = product_name[-1:]
print(last_letter)

order_code = "2026-SEOUL-042"
year = order_code[0:4]
print(year)

city = order_code[5:10]
print(city) 

order_number = order_code[-3:]
print(order_number)

file_name = "python_study_note.txt"

extension = file_name[-3:] 
print(extension)


message = "지금 배가 고픈거 같기도 한것같은데"
print(message[0:3])
print(message[-3:])
print(message[6:9])


name = "백성현"
greeting = "님 안녕하세요"
print(name + greeting)


```

```python
# 문자열 포맷팅
name = "김병주"
year = 2026
f_greeting = f"{name}님 안녕하세요! {year}년 압니다.".format(name=name,year=year)

print(f_greeting)
```

```python
# 문자열 포맷팅
name = "김병주"
year = 2026
f_greeting = "%s님 안녕하세요! %d년 입니다." % (name, year)

print(f_greeting)
```

### 문자열 메소드
- 문자열에는 자주 쓰는 도구가 많이 있습니다.
- 메서드 : .을 찍어서 사용합니다.
- 원본은 바뀌지 않고 새 문자열 을 돌려주는
- strip, upper/lower, replace, split, len 등등

```python
raw_text = " I Love you  "
print(raw_text.strip())
```

```python
print(raw_text.rstrip())
print(raw_text.lstrip())
```

```python
text = "2026-08-31"
text_blank = "2026 08 31"
text_slash = "2026/08/31"

print("원본",text.split("-"))
print("공백",text_blank.split(" "))
print("슬래시",text_slash.split("/"))
```

```python
student_info = "이현근/27/광진구"
name, age, gu = student_info.split("/")

print(name)
print(age)
print(gu)
```

```python
# 문자열 변경하기
student_info.replace("/","-")
```

```python
student_info = student_info.replace("/","-")
print(student_info)
```

```python
# 문자열 개수 구하기
slash_count = student_info.count("-")
slash_count
```

```python
# 문자열 길이 구하기
length = len(student_info)
length
```
