딕셔너리 - 키 와 값 구분

        -    딕셔너리 는 키와 밸류로 구성되어 있다.                

        -    key : value 연결되어 있어서 key 로 그 value 를 가져올 수 있다.

            -    예) 리스트는 순서로 값을 찾는데 -> 점수[0], 점수[1] -> 누군가의 점수가 궁금하다

                        -> 그 누군가의 index 번호를 알아야 하는데 -> 귀찮다. 민수점수, 택한점수, 수림점수 ...

    이름표 라고 생각하면 된다.

{} 에 key:value 나타냅니다.

딕셔너리[key] 으로 그 값을 가져올 수 있습니다.

```python
ai_family_dic = {"강기현":"01", "송수림":"78","이현근":"00"}
print(ai_family_dic)


ai_family_dic["강기현"]
```

```python
ai_family_dic["송수림"]
```

```python
ai_family_dic = {"강기현":"01", "송수림":"78","이현근":"00","김병주":["붉은사막","팰월드"]}
```

```python
ai_family_dic['김병주']
```

```python
ai_family_dic = {"강기현":"01", 
                 "송수림":"78",
                 "이현근":"00",
                 "김병주":["붉은사막","팰월드"],
                 "김수연":{"취미" : "조승우덕질","출생년도": "00"}
                 }
```

```python
ai_family_dic['김수연']
```

```python
ai_family_dic['김수연']['취미']
```

```python
# 딕셔너리 요소 추가
ai_family_dic["유종혁"] = {"취미":["게임","보드게임"],"출생년도":"89"}
print(ai_family_dic)
```

```python
ai_family_dic["백성현"] = {"취미":["유튜브음악"],"출생년도":"79"}
ai_family_dic["주승우"] = {"취미":["음악듣기"],"출생년도":"01"}
print(ai_family_dic)
```

```python
test_dic = {"게임":["붉은사막","팰월드","슈퍼마리오"],"교통수단":["버스","자전거"]}
print(test_dic)
print(test_dic["게임"])
```
