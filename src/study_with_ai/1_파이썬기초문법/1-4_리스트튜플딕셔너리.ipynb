```python
# 숫자로 된 리스트를 쉽게 만드는 방법을 배워보자
nums = list(range(5))
nums
```

```python
nums = list(range(4,15,3))
nums
```

```python
# 리스트 수정
nums[2] = 20
nums
```

```python
nums[1:3] = [30, 40] 
nums
```

```python
nums[1:3] = [40,60,50] 
nums
```

```python
### 리스트 삭제
nums = [10,20,30,40,50]
nums
```

```python
nums = list(range(10,51,10))
nums
```

```python
# del 요소 지우기
del nums[3]
nums
```

```python
# remove 로 지우기 - 일치하는 것을 지우기
nums.remove(10)
nums
```

```python
nums = [10, 20, 30, 30,40, 50]
nums.remove(30)
```

```python
nums = [10, 20, 30, 70, 30,40, 50]
nums.remove(30)
nums
```

```python
# pop 으로 지우기
nums.pop()
nums
```

```python
# revers, sort
nums.reverse()
nums
```
