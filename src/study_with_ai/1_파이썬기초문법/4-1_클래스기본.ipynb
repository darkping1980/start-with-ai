```python
class Human2:
    name = ""
    age = 0

    # 메소드 정의 
    def introduce(self):
        print(f"이름은 {self.name} 입니다.")
        print(f"나이는 {self.age} 입니다.")
```

```python
yth = Human2()
yth.name = "윤택한"
yth.age = 35

yth.__dict__
```

```python
yth.introduce()
```

```python
# 클래스 생성자 - 초기값 입력
class Book:
    # 초기에 반드시 있어야 하는 항목이 있을 경우
    def __init__(self,title,author):
        self.title = title      # 입력한 title 을 self.title 의 속성으로 저장
        self.author = author    # 입력받은 저자 

    # 메소드 추가 - 간단한 책소개 기능 추가
    def show_info(self):
        print(f"""
이 책의 제목은 {self.title} 이고
저자는 {self.author} 입니다.                                                                                   
많이 애용해주세요        
""")      
        
```

```python
my_book = Book("ComfyUI 독학노트","김한호, 김홍길, 권구영, 윤택한, 최태온")
```

```python
my_book.show_info()
```
