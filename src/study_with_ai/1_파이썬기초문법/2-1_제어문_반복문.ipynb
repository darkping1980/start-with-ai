### 반복문
- 같은 일을 반복할 수 있는게 프로그램의 특징
- 예) 인사를 100 출력
- 이런 반복되는 작업을 쉽게 할 수 있는게 반복문 코드다


#### 1. while
- while 조건 : 실행문
- 조건이 참일 경우 계속 실행되는 코드

```python
# 10 까지 숫자 출력
count = 1
while( count <= 10 ):
    print(count)
    count = count + 1
    

```

```python
# 연습: 10 에서 시작해서 1 씩 줄여서 0 보다 클동안 출력해보기
count = 10

while ( 0 < count ):
    print(count)
    count = count - 1
```

```python
# 인사 100번 시키기
count = 100

while count > 0:
    print(count,"안녕하세요")
    count = count - 1        
```

```python
# 인사 100번 시키기
count = 0

while count < 10:
    print(count,"횟수")
    count = count + 1     

    if( (count % 5) == 0 ):
        break;

```

```python
# 인사 100번 시키기
count = 0

while True:
    
    count = count + 1     
    print(count)

    if count == 7:
        break
```

```python
count = 0

while count < 10:
    count += 1
    
    if count % 2 == 0:
        continue            
    print(count)    
```

####  for  문

```python
num_list = [1,2,3,4,5]

for i in num_list:
    # print(i)
    # 다른작업
    print(i*5)    
```

```python
people = ["백성현","송수림","이현근","김수연","김병주"]

for name in people:
    print(f"{name} 님 안녕하세요")
```

```python
# 딕셔너리
data = {"백성현":1,"송수림":2,"이현근":3}

for i in data:
    print(i)
```

```python
for i in range(10):
    print("안녕하세요")
```

```python
for i in "안녕하세요":
    print(i)        
```

```python
data = ["주승우","유정혁","김수연","김병주","이현근"]

# index 넘버도 같이 출력 할수 있을까?
for name in data:
    print(name)
```

```python
data = ["주승우","유정혁","김수연","김병주","이현근"]

# index 넘버도 같이 출력 할수 있을까?
for i in range(len(data)):
    print(i, data[i])
```

```python
# enumberate'
for i, name in enumerate(data):
    print(i, name)
```

3. 리스트컴프리핸션

- list 를 만들때 for 문으로 정리한 데이터가 필요할 수 있는데 
- 짧은 한줄로 표현하고 싶을때

```python
# 1부터 5까지의 제곱을 리스트로 만들려고 한다면
# 1,4,9, 16,25

data = []   # 1. 데이터를 담을 빈 리스트를 만든다.

for i in range(1,6):
     print(i ** 2)
     data.append(i ** 2)

print()
print(data)     
```

```python
# 위 코드를 한줄로 나타내는게 -> 리스트 컴프리핸션
data_com = [ i**2 for i in range(1,6)]
data_com
```
