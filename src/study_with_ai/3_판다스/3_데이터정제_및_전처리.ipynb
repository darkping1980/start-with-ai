```python
df = pd.read_csv("../data/3-1_students.csv")
df
```

```python
# 결측치를 평균으로 채우기
korean_score = df['국어'].mean()
df['국어'].fillna(korean_score)
```

```python
df['국어'] = df['국어'].fillna(korean_score)
df
```

```python
df['영어'] = df['영어'].fillna(df['영어'].mean())
df
```

### 4. 중복 데이터 처리

### 6. 타입을 고치기
- 숫자여야 할 부분이 아닌 값들은 -> NaN 으로 바꾸기
- pd.to_numeric()

```python
pd.to_numeric( df['수학'],errors="coerce")
```

```python
df['수학'] = pd.to_numeric( df['수학'],errors="coerce")
df
```

```python
df['수학'] = df['수학'].fillna(df['영어'].mean())
df
```
