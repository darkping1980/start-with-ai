### 데이터프레임(DataFrame) : 표 -> 대부분의 데이터를 다루는 형태
- 엑셀처럼 행 과 열로 된 표를 다루는 가장 기본 형태
- 수천, 수만줄의 데이터도 코드 몇줄로 다루는게 가능하다
- 데이터 분석이라는 것을 해보고 싶다 -> 꼭 해야 합니다.

#### 1. Series - 한줄짜리 데이터
- 리스트와 비슷하지만, 각 값마다 인덱스가 붙습니다.
- 표의 한 열을 나타낸다.

```python
# 실습
import pandas as pd

data = [81,72,33,94] # 수학점수
data
```

```python
data_series = pd.Series(data)
data_series
```

```python
en_data = [92,88,67,54]
en_series = pd.Series(en_data)
en_series
```
