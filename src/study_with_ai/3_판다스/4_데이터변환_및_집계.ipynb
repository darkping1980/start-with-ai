### 3. 어떤 특정 그룹별로 한번 집계를 해보고 싶다(통계계산)
- groupby
- 공원 종류별로 묶어서, 개수가 몇개인지, 합계는 몇인지, 평균 면적은 얼마인지
- 어떤 종류의 공원이 많은지                                                                                                       

```python
# 공원면적이 0인 값, 혹은 얼마 이하인 값들은 제거해야 할 것 같다. -> 아직 처리는 안함
clean_df.groupby("공원구분")["공원면적"].count()
```

```python
summary = clean_df.groupby("공원구분")["공원면적"].agg(["count","sum","mean","median"])
summary 
```

```python
summary.sort_values("count",ascending=False)
```

####  인사이트
- 어린이공원, 소공원은 개수가 많지만 크기가 작다.
- 근린공원은 평균 면적이 크다.  전체 공원면적이 크다.
- 개수 1 위는 어린이공원이지만, 전체면적으로 보면 근린공원이 주된 공원이라고 할수도 있다.
- 근린공원위주로 가야 한다.

#### 4. 교차표 만들기 - pivot_table
- 어떤 두 컬럼(범주) 를 가로,세로로 교차시켜서 요약하는 
- 관리기관 * 공원구분의 평균 면적을 같이 보려고 할 때 

```python
table_df = clean_df.pivot_table(index="관리기관명", 
                                columns="공원구분", 
                                values="공원명",
                                aggfunc="count",
                                fill_value=0
                                )
table_df
```
