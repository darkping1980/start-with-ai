### 6. 여러 조건 - &(그리고),|(또는) 
- 조건이 여러개면 &, | 로 적절하게
- 각 조건은 소괄호로 감싸야합니다.

```python
# A 반이면서 국어가 85점이상 학생
df[ (df["반"] == 'A') & (df['국어'] >= 85 ) ]
```

```python
df.head
```

```python
df[(df['성별'] == '여') & ( df['영어'] > 90)]
```

```python
df[(df['성별'] == '여') | ( df['국어'] > 80)]
```

```python
df.head()
```

```python
# 영어가 90점 이상이거나, 국어가 86점 이상인 사람은?
df[ (df["국어"] >= 86) | ( df["영어"] >= 90 )  ]
```
